```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d, CubicSpline
from scipy.integrate import simpson
from scipy.optimize import minimize
import camb
import os

# ============================================================
# 1. Template generator (your Hamman‑2010 method)
# ============================================================

def generate_templates(force_recompute=False, save_path=None):
    """
    Returns:
        k_uniform : 1D array, uniform in linear k
        P_wiggle  : 1D array, linear P(k) with BAO wiggles
        P_nowiggle: 1D array, no‑wiggle P(k)
        r_array   : 1D array, separation grid (for diagnostic xi)
        xi_wiggle, xi_nowiggle : correlation functions (unused in fitter)
    """
    # Build a unique filename to cache the result
    if save_path is None:
        save_path = 'templates_hamman.npz'

    if not force_recompute and os.path.exists(save_path):
        data = np.load(save_path)
        k_uniform = data['k']
        P_wiggle = data['P_w']
        P_nowiggle = data['P_nw']
        r_array = data['r']
        xi_wiggle = data['xi_w']
        xi_nowiggle = data['xi_nw']
        print(f"Loaded templates from {save_path}")
        return k_uniform, P_wiggle, P_nowiggle, r_array, xi_wiggle, xi_nowiggle

    # --- Parameters ---
    k_min_BAO = 0.04
    k_max_BAO = 0.8
    percentile_d2 = 50
    n = 11                     # N = 2^n uniform grid points
    N = 2**n

    # Cosmology
    H0 = 67.7
    omega_b = 0.022
    omega_cdm = 0.12
    ns = 0.96
    As = 2.1e-9

    # CAMB
    pars = camb.CAMBparams()
    pars.set_cosmology(H0=H0, ombh2=omega_b, omch2=omega_cdm)
    pars.InitPower.set_params(As=As, ns=ns)
    k_min_hMpc = 1e-3
    k_max_hMpc = 10.0
    pars.set_matter_power(redshifts=[0], kmax=k_max_hMpc)
    results = camb.get_results(pars)
    kh, z, pk = results.get_matter_power_spectrum(minkh=k_min_hMpc, maxkh=k_max_hMpc, npoints=10000)
    Plin = pk[0]

    # Uniform k grid and interpolate f(k)=ln(k*P)
    k_uniform = np.linspace(k_min_hMpc, k_max_hMpc, N)
    interp_lnkP = interp1d(kh, np.log(kh * Plin), kind='linear', fill_value='extrapolate')
    f = interp_lnkP(k_uniform)

    # Split even/odd and fit splines
    even_mask = np.arange(N) % 2 == 0
    odd_mask = ~even_mask
    spline_even = CubicSpline(k_uniform[even_mask], f[even_mask], bc_type='natural')
    spline_odd  = CubicSpline(k_uniform[odd_mask],  f[odd_mask],  bc_type='natural')

    # Curvature thresholds
    bao_mask = (k_uniform >= k_min_BAO) & (k_uniform <= k_max_BAO)
    d2_even = spline_even.derivative(2)(k_uniform)
    d2_odd  = spline_odd.derivative(2)(k_uniform)
    thresh_even = np.percentile(np.abs(d2_even[bao_mask]), percentile_d2)
    thresh_odd  = np.percentile(np.abs(d2_odd[bao_mask]),  percentile_d2)

    # Find and remove bad points inside BAO range
    bad = bao_mask & (np.abs(d2_even) > thresh_even) & (np.abs(d2_odd) > thresh_odd)
    k_good = k_uniform[~bad]
    f_good = f[~bad]
    spline_clean = CubicSpline(k_good, f_good, bc_type='natural')
    f_clean = f.copy()
    f_clean[bao_mask] = spline_clean(k_uniform[bao_mask])

    # Reconstruct P(k)
    kP_wiggle = np.exp(f)
    kP_nowiggle = np.exp(f_clean)
    P_wiggle = kP_wiggle / k_uniform
    P_nowiggle = kP_nowiggle / k_uniform

    # Compute diagnostic correlation functions (for plotting/sanity)
    r_array = np.logspace(np.log10(1.0), np.log10(200.0), 300)
    factor = 1.0 / (2.0 * np.pi**2)
    xi_wiggle = np.zeros_like(r_array)
    xi_nowiggle = np.zeros_like(r_array)
    for i, r in enumerate(r_array):
        integrand_w  = kP_wiggle * np.sin(k_uniform * r)
        integrand_nw = kP_nowiggle * np.sin(k_uniform * r)
        xi_wiggle[i]   = (factor / r) * np.trapezoid(integrand_w, k_uniform)
        xi_nowiggle[i] = (factor / r) * np.trapezoid(integrand_nw, k_uniform)

    # Save for future use
    np.savez(save_path,
             k=k_uniform, P_w=P_wiggle, P_nw=P_nowiggle,
             r=r_array, xi_w=xi_wiggle, xi_nw=xi_nowiggle)
    print(f"Templates computed and saved to {save_path}")
    return k_uniform, P_wiggle, P_nowiggle, r_array, xi_wiggle, xi_nowiggle


# ============================================================
# 2. BAO model – damped in k‑space, Fourier transformed to xi(s)
# ============================================================

def xi_damped(s, k, P_wiggle, P_nowiggle, Sigma, interp=False):
    """
    Compute xi(s) from damped power spectrum:
       P(k) = P_nw(k) + (P_w(k)-P_nw(k)) * exp(-k² Sigma²/2)
    using Simpson integration.
    s can be a scalar or array.
    If interp=True, s values outside [1, 200] Mpc/h are extrapolated linearly.
    """
    # Damped power spectrum
    damping = np.exp(-k**2 * Sigma**2 / 2.0)
    P_damped = P_nowiggle + (P_wiggle - P_nowiggle) * damping

    # Prefactor: xi(s) = (1/(2π² s)) ∫ k P(k) sin(k s) dk
    s_arr = np.atleast_1d(s).astype(float)
    xi = np.zeros_like(s_arr)
    factor = 1.0 / (2.0 * np.pi**2)
    for i, si in enumerate(s_arr):
        if si == 0:
            # limit: xi(0) = (1/(2π²)) ∫ k² P(k) dk
            integrand = P_damped * k**2
            xi[i] = factor * simpson(integrand, x=k)
        else:
            integrand = k * P_damped * np.sin(k * si)
            xi[i] = (factor / si) * simpson(integrand, x=k)
    if np.ndim(s) == 0:
        return float(xi[0])
    return xi


def xi_model(params, s, k, P_wiggle, P_nowiggle):
    """
    Full BAO model:
        xi_model(s) = B² * xi_damped(α s; Σ) + A0 + A1/s + A2/s²
    """
    alpha, B, Sigma, A0, A1, A2 = params
    s_scaled = alpha * s
    xi_temp = xi_damped(s_scaled, k, P_wiggle, P_nowiggle, Sigma)
    return B**2 * xi_temp + A0 + A1/s + A2/s**2


# ============================================================
# 3. Chi² and fitting
# ============================================================

def chi2(params, s_data, xi_data, k, P_wiggle, P_nowiggle):
    model = xi_model(params, s_data, k, P_wiggle, P_nowiggle)
    return np.sum((xi_data - model)**2)


def fit_bao(s_data, xi_data, k, P_wiggle, P_nowiggle, fit_range=(80,120)):
    mask = (s_data >= fit_range[0]) & (s_data <= fit_range[1])
    s_fit = s_data[mask]
    xi_fit = xi_data[mask]

    p0 = [1.0, 1.0, 6.0, 0.0, 0.0, 0.0]   # alpha, B, Sigma, A0, A1, A2
    bounds = [(0.8, 1.2), (0.0, None), (0.0, 20.0),
              (None, None), (None, None), (None, None)]

    result = minimize(chi2, p0, args=(s_fit, xi_fit, k, P_wiggle, P_nowiggle),
                      method='L-BFGS-B', bounds=bounds)
    return result


# ============================================================
# 4. Plotting
# ============================================================

def plot_fit(s_data, xi_data, result, k, P_wiggle, P_nowiggle):
    xi_best = xi_model(result.x, s_data, k, P_wiggle, P_nowiggle)
    residual = xi_data - xi_best

    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8,6), sharex=True)
    ax1.plot(s_data, s_data**2 * xi_data, 'k.', label='Data')
    ax1.plot(s_data, s_data**2 * xi_best, 'r-', label='Best fit')
    ax1.set_ylabel(r'$s^2\xi(s)$')
    ax1.legend()

    ax2.plot(s_data, s_data**2 * residual, 'k.')
    ax2.axhline(0, color='gray', linestyle='--')
    ax2.set_xlabel(r'$s\ [\mathrm{Mpc}/h]$')
    ax2.set_ylabel(r'$s^2\Delta\xi$')
    plt.tight_layout()
    return fig


# ============================================================
# 5. Load monopole data
# ============================================================

def load_monopole(filename):
    data = np.load(filename)
    # Adjust keys if your files use different names
    return data['s'], data['xi0']


# ============================================================
# 6. High‑level runner
# ============================================================

def run_bao_fit(monopole_file, template_kwargs=None, fit_range=(80,120)):
    #k, P_w, P_nw, *_ = generate_templates(**(template_kwargs or {}))
    s, xi = load_monopole(monopole_file)
    
    k, P_w, P_nw, r, xi_w, xi_nw = generate_templates()
    # Check if the template xi_w(r) matches the r array
    plt.plot(r, r**2 * xi_w)
    plt.show()

    # Compute xi_damped at a single point
    xi_test = xi_damped(np.array([100.0]), k, P_w, P_nw, 6.0)
    print("xi_damped at s=100, Sigma=6:", xi_test)

    # Fit – also return the fit mask arrays
    mask = (s >= fit_range[0]) & (s <= fit_range[1])
    s_fit = s[mask]
    xi_fit = xi[mask]
    
    result = minimize(chi2, [1.0, 1.0, 6.0, 0,0,0],
                      args=(s_fit, xi_fit, k, P_w, P_nw),
                      method='L-BFGS-B',
                      bounds=[(0.8,1.2),(0,None),(0,20),(None,None),(None,None),(None,None)])
    
    chi2_min = result.fun
    dof = len(s_fit) - len(result.x)
    
    print("\n=== BAO FIT RESULTS ===")
    print(f"χ² = {chi2_min:.2f} / {dof}  (reduced χ² = {chi2_min/dof:.3f})")
    print(f"alpha  = {result.x[0]:.4f}")
    print(f"B      = {result.x[1]:.4f}")
    print(f"Sigma  = {result.x[2]:.4f} Mpc/h")
    print(f"A0, A1, A2 = {result.x[3]:.4f}, {result.x[4]:.4f}, {result.x[5]:.4f}")
    print("========================\n")
    
    fig = plot_fit(s, xi, result, k, P_w, P_nw)
    fig.suptitle(os.path.basename(monopole_file), fontsize=12)
    plt.show()
    return result


# ============================================================
# 7. Main – run the three fits
# ============================================================
if __name__ == "__main__":
    # Make sure the template file is in a reasonable location
    template_kwargs = dict(save_path='templates_hamman.npz', force_recompute=False)

    samples = {
        'Full Sample': '../data/monopoles/box/mag=-21.2_sep=1.0-150.0_binsep=2.0_full.npz',
        'Filament Galaxies': '../data/monopoles/box/mag=-21.2_sep=1.0-150.0_binsep=2.0_0.0-d_mathrmfilleq3.0.npz',
        'Non-filament Galaxies': '../data/monopoles/box/mag=-21.2_sep=1.0-150.0_binsep=2.0_8.0-d_mathrmfilleq100.0.npz'
    }

    for label, fname in samples.items():
        print(f"\n{'#'*40}\n{label}\n{'#'*40}")
        run_bao_fit(monopole_file=fname, template_kwargs=template_kwargs)
```
